## 1. DATA CLEANING GUIDELINES FOR train.csv

```python
train_df = spark.read.format("csv").option("header", True).option("inferSchema", True).load("/FileStore/walmart-sales-forecast/train.csv")
train_df = train_df.toDF(*[c.lower() for c in train_df.columns])
train_df.display()
```

```python
train_df = train_df.withColumn("date", to_timestamp(col("date"), "YYYY-MM-dd'T'HH:mm:ss"))
train_df.display()
```

```python
# If Weekly_Sales is NULL → fill with median sales of that department
median_sale = train_df.groupBy("dept").agg(percentile_approx("weekly_sales", "0.5").alias("median_sales"))
train_df = train_df.join(median_sale, on = "dept", how = "left")
final_train_df = train_df.withColumn("weekly_sales", when(col("weekly_sales").isNull(), col("median_sales")).otherwise(col("weekly_sales"))).drop(col("median_sales"))
final_train_df.display()
```

```python
final_train_df = final_train_df.filter(col("weekly_sales") >= 0)
final_train_df.display()
```

```python
deduped_train_df = final_train_df.dropDuplicates(["store", "dept", "date"])
deduped_train_df.display() 
```
